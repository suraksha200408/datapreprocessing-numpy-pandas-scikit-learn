```python
import pandas as pd
df = pd.DataFrame({
    "City": ["Bengaluru", "Mysuru", "Bengaluru", "Mysuru",
             "Bengaluru", "Mysuru", "Bengaluru", "Mysuru"],
    "Category": ["Electronics", "Electronics", "Clothing", "Clothing",
                 "Electronics", "Clothing", "Clothing", "Electronics"],
    "Sales": [50000, 30000, 15000, 20000, 45000, 25000, 18000, 35000],
    "Profit": [8000, 4000, 3000, 5000, 7000, 6000, 3500, 4500]
})
pd.pivot_table(df,index="City",columns="Category",values="Sales", aggfunc="sum")
```

```python
pd.pivot_table(df,index="City",columns="Category",values="Profit", aggfunc="mean")
```

```python
result = pd.pivot_table(
    df,
    index="City",
    columns="Category",
    values=["Sales", "Profit"],
    aggfunc={"Sales": "sum", "Profit": "mean"}
)
print(result)
```
