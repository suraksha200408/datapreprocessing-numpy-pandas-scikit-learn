```python
import pandas as pd
import numpy as np

data = {
    "Employee": ["Asha", "Rahul", "Meera", "Asha", "Kiran",
                 "Rahul", "Divya", "Meera", "Vijay"],
    "Department": ["IT", "HR", "IT", "IT", "Finance",
                   "HR", "Finance", "IT", "HR"],
    "Salary": [60000, 45000, np.nan, 60000, 55000,
               48000, 70000, 65000, np.nan],
    "Experience": [4, 2, 5, 4, np.nan, 3, 7, 6, 1],
    "Performance": [90, 75, 88, 90, 82, 78, 95, 92, 70]
}

df = pd.DataFrame(data)

df=df.drop_duplicates(keep="first")
df["Salary"]=df["Salary"].fillna(df["Salary"].median())
df["Experience"]=df["Experience"].fillna(0)
df["Bonus"]=df.apply( lambda mark:mark["Salary"]+(10/100) if mark["Performance"]>=90 else mark["Salary"]+(5/100), axis=1 )
df["Total_Pay"]=df["Salary"]+df["Bonus"]
df=df.loc[(df["Total_Pay"]>60000) & (df["Experience"]>=4)]
df= df.sort_values(by="Total_Pay",ascending=False)
print(df.loc[:,["Employee", "Department", "Experience", "Performance", "Total_Pay"]])
```
