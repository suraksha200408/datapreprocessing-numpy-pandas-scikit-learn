```python
import pandas as pd
import numpy as np

data = {
    "Employee": ["Anu", "Bharath", "Charan", "Deepa", "Esha", "Farhan"],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "IT"],
    "Salary": [45000, 38000, 62000, 55000, 42000, 75000],
    "Performance": [85, 65, 92, 78, 88, 96],
    "Status": ["active", "Active", "inactive", "active", "INACTIVE", "active"]
}

df = pd.DataFrame(data)
df["Performance_Level"]= df["Performance"].apply(lambda scores:"Excellent" if scores>=90 else "Good" if (scores>=80) & (scores<90)  else "needs improvement")
df["Bonus"]=df.apply( lambda b:b["Salary"]*0.10 if b["Performance"]>=90 else b["Salary"]*0.05, axis=1)
df["department_Full"]=df["Department"].map({ "IT": "Information technology", "HR": "Human Resources", "Finance" :"Finance"})
df["Status"]=df["Status"].replace({ "active":"Active", "inactive":"Inactive"})
print(df)
employee=df[(df["Bonus"]>3000) & (df["Performance"]>85) ] 
employee=employee.sort_values(by="Bonus", ascending= False)
print(employee.loc[:,["Employee", "Department", "Performance", "Bonus"]])
```

```python
import pandas as pd
import numpy as np

data = {
    "Crop": ["rice", "WHEAT", "Maize", "rice", "RAGI", "wheat"],
    "Market": ["Kolar", "Mandya", "Kolar", "Mysuru", "Mandya", "Kolar"],
    "Price_Per_Quintal": [2450, 2200, 1850, 2600, 3200, 2100],
    "Demand": [850, 620, 450, 920, 300, 700],
    "Quality_Code": ["A", "B", "A", "C", "B", "X"]
}

df = pd.DataFrame(data)
df["Crop"]=df["Crop"].replace({ "rice":"Rice" , "WHEAT":"Wheat"})
df["qty_desc"]=df["Quality_Code"].map({ "A":"Excellent", "B":"Good", "C": "satisfied", "X": "Unclassified"})
df["price_fee"]= df["Price_Per_Quintal"].apply(lambda price:price*0.02 if price>=2500 else price*0.01)
print(df)
df["Revenue_Estimate"]=df["Price_Per_Quintal"]*df["Demand"]
df["Price_Category"]=df["Price_Per_Quintal"].apply(lambda rev:"High" if rev>=2500 else "Medium" if (rev>=2000 & rev<2500) else "low" )
df=df.loc[df["Revenue_Estimate"]>1500000, ["Crop","Market","Revenue_Estimate","Price_Category"]]
print(df)
```
