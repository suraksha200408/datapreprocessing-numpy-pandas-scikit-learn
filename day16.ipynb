```python
"""Merge the two DataFrames to retain only orders belonging to existing customers.
Calculate the total order amount for each customer, including only Delivered orders.
Return a DataFrame containing Customer_ID, Name, City, and Total_Sales, sorted by Total_Sales in descending order."""
import pandas as pd

customers = pd.DataFrame({
    "Customer_ID": [101, 102, 103, 104, 105],
    "Name": ["Anu", "Rahul", "Sneha", "Kiran", "Divya"],
    "City": ["Bengaluru", "Mysuru", "Bengaluru", "Hubli", "Mysuru"],
    "Membership": ["Gold", "Silver", "Gold", "Bronze", "Silver"]
})

orders = pd.DataFrame({
    "Order_ID": [501, 502, 503, 504, 505, 506, 507],
    "Customer_ID": [101, 102, 101, 103, 106, 104, 103],
    "Amount": [5000, 2500, 3000, 4500, 1800, 1200, 3500],
    "Status": ["Delivered", "Pending", "Delivered",
               "Delivered", "Cancelled", "Delivered", "Pending"]
})
delivered = orders[orders["Status"] == "Delivered"]

result = pd.merge(
    customers,
    delivered,
    on="Customer_ID",
    how="inner"
)
result["Total_Sales"] = (
    result.groupby("Customer_ID")["Amount"].transform("sum")
)

result = result[
    ["Customer_ID", "Name", "City", "Total_Sales"]
].drop_duplicates()

result = result.sort_values(
    by="Total_Sales", ascending=False
)

print(result)
```

```python
"""Join the two DataFrames using their indexes, retaining every employee record.
Identify employees whose performance rating is at least 4.0.
Create a final DataFrame containing Name, Department, Salary, Rating, and Projects_Completed.
Fill missing performance ratings with 0 and missing project counts with 0."""

employees = pd.DataFrame({
    "Name": ["Asha", "Bharath", "Charan", "Deepa", "Eshan"],
    "Department": ["IT", "HR", "IT", "Finance", "HR"],
    "Salary": [60000, 45000, 70000, 65000, 48000]
}, index=[201, 202, 203, 204, 205])

performance = pd.DataFrame({
    "Rating": [4.5, 3.2, 4.8, 4.1, 3.7],
    "Projects_Completed": [8, 4, 10, 7, 5]
}, index=[201, 203, 204, 205, 206])
result=employees.join(performance)

result=result.fillna(0)
result=result[result["Rating"]>=4.0]

result=result.fillna(0)
print(result)
```

```python
"""1. Combine all three DataFrames vertically, preserving every row and creating a fresh continuous index.
2. Ensure all columns from all months are retained.
3. Replace missing Units_Sold values with 0.
4. Replace missing Discount values with 0.
5. Calculate total revenue for each product and return the result sorted from highest to lowest revenue."""

jan_sales = pd.DataFrame({
    "Product": ["Laptop", "Phone", "Tablet"],
    "Revenue": [150000, 90000, 40000],
    "Units_Sold": [5, 10, 8]
})

feb_sales = pd.DataFrame({
    "Product": ["Laptop", "Phone", "Headphones"],
    "Revenue": [180000, 110000, 25000],
    "Discount": [5000, 3000, 1000]
})

mar_sales = pd.DataFrame({
    "Product": ["Tablet", "Laptop", "Headphones"],
    "Revenue": [60000, 200000, 35000],
    "Units_Sold": [12, 6, 15],
    "Discount": [2000, 7000, 1500]
})

all_sales=pd.concat([jan_sales,feb_sales,mar_sales],ignore_index=True)
all_sales=all_sales.fillna(0)
print(all_sales)
all_sales=all_sales.groupby("Product")["Revenue"].sum()
print(all_sales)
print(all_sales.sort_values(ascending=False))
```
