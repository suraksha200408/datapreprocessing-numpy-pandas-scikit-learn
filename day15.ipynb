```python
import pandas as pd

df = pd.DataFrame({
    "Region": ["South", "North", "South", "East", "North",
               "East", "South", "North", "East", "South"],
    "Category": ["Electronics", "Clothing", "Clothing", "Electronics",
                 "Electronics", "Clothing", "Electronics", "Clothing",
                 "Electronics", "Clothing"],
    "Sales": [25000, 12000, 8000, 30000, 18000,
              9000, 15000, 14000, 22000, 11000],
    "Profit": [5000, 3000, 1500, 7000, 4000,
               2000, 2500, 3500, 6000, 1800],
    "Units_Sold": [5, 12, 8, 6, 4, 9, 3, 14, 5, 11]
})

df.groupby(["Region", "Category"], as_index=False).agg(Total_Sales=("Sales","sum"),Avg_Profit=("Profit", "mean"), Total_Units=("Units_Sold", "sum"))
```

```python
result = df.groupby(
    ["Region", "Category"],
    as_index=False
).agg(
    Total_Sales=("Sales", "sum"),
    Avg_Profit=("Profit", "mean"),
    Total_Units=("Units_Sold", "sum")
)

result = result[
    (result["Total_Sales"] > 30000) &
    (result["Avg_Profit"] > 3000) &
    (result["Total_Units"] >= 10)
]

print(result)
```

```python
df["Region_Total_Sales"] = (
    df.groupby("Region")["Sales"].transform("sum")
)

df["Sales_Contribution_Pct"] = (
    df["Sales"] / df["Region_Total_Sales"] * 100
)

print(df)
```
