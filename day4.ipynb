```python
import numpy as np

sensor_data = np.array([
    [25.5, 40.2, 100.1],
    [26.1, 41.0, 102.5],
    [np.nan, 39.8, 101.2],
    [24.8, np.inf, 99.5],
    [27.2, 42.1, 103.4],
    [np.inf, 38.5, 98.9],
    [25.9, 40.7, 100.8]
])
#Identify all rows containing invalid values.
rows=~np.isfinite(sensor_data)
finite=np.any(rows==True, axis=1)
print("containing invalid values",sensor_data[finite,:])
#Identify all rows containing only valid values.
finite=np.all(rows==False, axis=1)
print("containing only valid values.",sensor_data[finite,:])
#Create a new array containing only valid rows.
new_array=sensor_data[finite,:]
#Calculate the mean of each sensor column using only valid rows.
print("mean of each sensor column",np.mean(new_array,axis=0))
```

```python
import numpy as np

market_prices = np.array([
    [2500, 2700, 2600, 2800],
    [2300, 2400, 2350, 2500],
    [3100, 3050, 3200, 3300],
    [1800, 1950, 1900, 1850],
    [2900, 3000, 3100, 3050],
    [2200, 2100, 2250, 2300]
])
#Highest average market
high=np.mean(market_prices,axis=1)
print(np.max(high))
#Identify markets whose average price is greater than 2,600.
print(high[high>2600])
#Calculate the price variation (maximum minus minimum) for each market.
hightest=np.max(market_prices,axis=1)
lowest=np.min(market_prices,axis=1)
prices=hightest-lowest
print(prices)
#Find all prices across the dataset that are greater than 3,000.
print(market_prices[market_prices>3000])
```
