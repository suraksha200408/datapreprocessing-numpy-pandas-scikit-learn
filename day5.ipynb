```python
import numpy as np

market_prices = np.array([
    2200, 1800, 2500, 2000, 3000,
    2700, 1900, 3200, 2400, 2800,
    2100, 2600, 3100, 2300, 2900
])

# 1. Set a fixed seed
np.random.seed(42)

# 2. Select 6 unique prices without replacement
selected_prices = np.random.choice(
    market_prices,
    size=6,
    replace=False
)

print("Selected prices:", selected_prices)

# 3. Calculate mean and standard deviation
print("Mean:", np.mean(selected_prices))
print("Standard deviation:", np.std(selected_prices))

# 4. Verify that no selected price is repeated
print("All prices are unique:",
      len(selected_prices) == len(np.unique(selected_prices)))
```

```python
import numpy as np

features = np.array([
    [25.0, 60.0, 101.2],
    [28.0, 65.0, 100.8],
    [30.0, 55.0, 102.0],
    [22.0, 70.0, 99.5],
    [27.0, 62.0, 101.5],
    [32.0, 58.0, 103.1],
    [24.0, 68.0, 100.1],
    [29.0, 63.0, 102.5]
])

# 1. Set seed and generate a permutation of row indices
np.random.seed(5)
indices = np.random.permutation(len(features))

# 2. Create a shuffled copy
shuffled_features = features[indices].copy()

# 3. Print original and shuffled datasets
print("Original dataset:")
print(features)

print("\nShuffled dataset:")
print(shuffled_features)

# 4. Verify original dataset remains unchanged
print("\nOriginal dataset unchanged:")
print(np.array_equal(features, np.array([
    [25.0, 60.0, 101.2],
    [28.0, 65.0, 100.8],
    [30.0, 55.0, 102.0],
    [22.0, 70.0, 99.5],
    [27.0, 62.0, 101.5],
    [32.0, 58.0, 103.1],
    [24.0, 68.0, 100.1],
    [29.0, 63.0, 102.5]
])))
```
