```python
"""Perform all the following operations:
1. Create a Bonus_Marks column equal to 5 for students with Marks of at least 80; otherwise, assign 2.
2. Create Final_Marks by adding Bonus_Marks to Marks.
3. Select students whose Final_Marks are greater than 80.
4. Sort the selected students by Final_Marks in descending order.
5. Display only Student, Marks, Bonus_Marks, and Final_Marks."""

import pandas as pd

data = {
    "Student": ["Anu", "Bharath", "Charan", "Deepa", "Esha",
                "Farhan", "Gita", "Harish"],
    "Department": ["CSE", "ISE", "CSE", "ECE", "ISE",
                   "ECE", "CSE", "ISE"],
    "Study_Hours": [4, 7, 5, 3, 8, 6, 9, 2],
    "Marks": [68, 88, 75, 55, 94, 82, 91, 45],
    "Attendance": [80, 95, 85, 70, 98, 90, 96, 60]
}

df = pd.DataFrame(data)
print(df.loc[(df["Marks"]>75) & (df["Attendance"]>=80) , ["Student" ,"Marks","Attendance"]])

df=df.sort_values(by=["Department", "Marks"],  ascending= [True, False])
print(df)
df.loc[df["Attendance"]<80, "Study_Hours"]+=2
print(df)
df["Bonus_Mark"]= df["Marks"].apply(lambda mark: 5 if mark>=80 else 2)
print(df)
df["Final_Marks"]=df["Bonus_Mark"]+df["Marks"]
df.loc[df["Final_Marks"] > 80, ["Student", "Marks", "Bonus_Mark", "Final_Marks"]]
df=df.sort_values(by= "Final_Marks", ascending=False)
print(df)
```
