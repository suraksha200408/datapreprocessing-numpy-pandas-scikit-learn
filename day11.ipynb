```python
"""Display the data types of every column.
Convert Price to numeric, replacing invalid values with missing values.
Convert Rating to numeric using the same approach.
Convert Date into datetime format.
After converting the columns, select rows where Price > 1900 and Rating >= 4.0, displaying Product, Price, Rating, and Date."""

import pandas as pd

data = {
    "Product": ["Rice", "Wheat", "Maize", "Ragi", "Barley"],
    "Price": ["2500", "1800", "unknown", "2300", "2000"],
    "Rating": ["4.5", "3.8", "4.2", "missing", "4.0"],
    "Date": ["2026-01-10", "2026-02-15", "2026-03-20",
             "2026-04-05", "2026-05-12"]
}

df = pd.DataFrame(data)
print(df.dtypes)
df["Price"]=pd.to_numeric(df["Price"], errors="coerce")
df["Rating"]=pd.to_numeric(df["Rating"], errors="coerce")
df["Date"]= pd.to_datetime(df["Date"], errors="coerce")
df=df.loc[(df["Price"] > 1900) & (df["Rating"] >= 4.0), ["Product", "Price", "Rating","Date"]]
print(df)
print(df.dtypes)
```
