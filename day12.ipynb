```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Area": [1200, 1500, np.nan, 1800, 2000, 1600, np.nan, 1400],
    "Bedrooms": [2, 3, 2, np.nan, 4, 3, 3, np.nan],
    "Location": ["Urban", "Rural", "Urban", "Urban", None, "Rural", "Rural", "Urban"],
    "Age": [5, np.nan, 10, 8, np.nan, 6, 12, 7],
    "Price": [4500000, 6000000, 4000000, 7500000, 9000000, 6200000, 5000000, 4800000]
})
count=df.isnull().sum()
print(count)
missing_percentage = (df.isnull().sum() / len(df)) * 100
print(missing_percentage)
df["Area"]=df["Area"].fillna(df["Area"].median())
df["Bedrooms"]=df["Bedrooms"].fillna(df["Bedrooms"].median())
df["Age"]=df["Age"].fillna(df["Age"].median())
df["Location"]=df["Location"].fillna(df["Location"].mode()[0])
clean_df=df
print(clean_df.isnull().sum())
```
