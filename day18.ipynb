```python
import pandas as pd

df = pd.DataFrame({
    "Date": [
        "2025-01-10", "2025-02-15", "2025-02-25",
        "2025-03-05", "2025-03-20", "2025-04-12",
        "2025-04-25", "2025-05-10"
    ],
    "Crop": [
        "Rice", "Wheat", "Rice", "Maize",
        "Wheat", "Rice", "Maize", "Wheat"
    ],
    "Price": [2500, 2200, 2600, 1800, 2300, 2700, 1900, 2400],
    "Quantity": [100, 150, 120, 200, 130, 110, 180, 140]
})

df["Date"] = pd.to_datetime(df["Date"])
result = df[
    (df["Date"] >= "2025-02-01") &
    (df["Date"] < "2025-03-01")
]

print(result)
```

```python
df["Month"] = df["Date"].dt.month_name()

monthly_avg = df.groupby("Month", as_index=False)["Price"].mean()

print(monthly_avg)
```

```python
df["Month"] = df["Date"].dt.month_name()

df["Revenue"] = df["Price"] * df["Quantity"]

monthly_crop_revenue = df.groupby(
    ["Month", "Crop"], as_index=False
)["Revenue"].sum()

print(monthly_crop_revenue)
```
