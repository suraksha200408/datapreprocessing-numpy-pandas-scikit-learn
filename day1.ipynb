```python
import numpy as np

data = np.array([
    [42, 120, 0.8, 210],
    [55, 145, 1.2, 250],
    [38, 110, 0.5, 195],
    [78, 190, 3.8, 340],
    [61, 160, 2.1, 285],
    [33, 95, 0.4, 175],
    [85, 205, 4.5, 390],
    [49, 135, 1.0, 230],
    [72, 180, 3.2, 320],
    [46, 125, 0.7, 215]
])

# Columns:
# Temperature, Pressure, Vibration, Power consumption

#Write code to print the number of dimensions, the shape, and the total number of elements in data.
print(data.ndim)
print(data.shape)
print(data.size)
```

```python
#Extract only the Temperature and Vibration columns, keeping every row. The result must remain a 2D array.

print(data[:,[0,2]])
```

```python
#Select rows at index 2 through index 6, excluding index 7. Keep every feature.

print(data[2:7,:])
```

```python
#Return complete rows where Temperature is greater than 60.
print(data[data[:,0]>60])
```

```python
#Return individual Vibration values that are below 1.0 or above 3.0.
print(data[:, 2][(data[:, 2] < 1.0) | (data[:, 2] > 3.0)])
```

```python
#Return complete rows where Temperature is greater than 50, Vibration is greater than 2.0, AND Power consumption is greater than 280.

print(data[(data[:,0]>50) & (data[:,2]>2.0) & (data[:,3]>280)])
```

```python
#Return complete rows where Temperature is above 75 OR Vibration is above 4.0. Each row should appear only once, even if both conditions are true.
print(data[(data[:,0]>75) | (data[:,2]>4.0) ])
```

```python
#Return complete rows where Pressure is between 120 and 180 inclusive, while Temperature is below 60.
print(data[
    (data[:, 1] >= 120) &
    (data[:, 1] <= 180) &
    (data[:, 0] < 60)
])
```

```python
#Create a new array in which Temperature is increased by 2, Pressure is multiplied by 1.1, Vibration remains unchanged, and Power consumption is decreased by 10. Do not modify the original data array.

# Create a copy so the original array is unchanged
new_data = data.copy()

# Increase Temperature by 2
new_data[:, 0] = new_data[:, 0] + 2

# Multiply Pressure by 1.1
new_data[:, 1] = new_data[:, 1] * 1.1

# Keep Vibration unchanged

# Decrease Power consumption by 10
new_data[:, 3] = new_data[:, 3] - 10

print("Original data:")
print(data)
```
