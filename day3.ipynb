```python
import numpy as np

data = np.array([
    [28.0, 120.0, 2500.0, 500.0],
    [32.0, np.nan, 2800.0, 600.0],
    [35.0, 80.0, np.inf, 450.0],
    [26.0, 150.0, 2200.0, 550.0],
    [30.0, 100.0, 2600.0, -np.inf]
])
#1.Count invalid values in each column.
#2.Create a mask for rows where every value is finite.
#3.Store the valid rows in clean_data and print its shape.

count=np.sum(~np.isfinite(data),axis=0)
print(count)

mask=np.all(np.isfinite(data), axis=1) 
clean_data=data[mask,:]
print(clean_data, clean_data.shape)
```

```python
import numpy as np

prices = np.array([
    1800, 2200, 2500, 4100,
    2300, 5000, 2100, 2600
])
#Select all prices greater than ₹3,000.
#Count how many prices exceed ₹3,000.
#Use np.any() to check whether any price exceeds ₹4,500.
#Use np.all() to check whether all prices are above ₹1,500.

print(prices[prices>3000],prices[prices>3000].size)

print(np.any(prices>4500))

print(np.all(prices>1500))

```

```python
import numpy as np

sensor_data = np.array([
    [25.0, 60.0, 101.2],
    [np.nan, 55.0, 100.8],
    [30.0, np.inf, 102.0],
    [28.0, 65.0, -np.inf],
    [27.0, 62.0, 101.5]
])
#Create a Boolean mask identifying all non-finite values.
#Count invalid values in each row.
#Replace every non-finite value with -1 using np.where().
#Verify that the cleaned array contains only finite values using np.all().

mask=np.isfinite(sensor_data)
print(mask)
s=np.sum(mask==False,axis=1)
print(s)
create=np.where(~np.isfinite(sensor_data),-1,sensor_data)
print(create)
print(np.all(create))
```
