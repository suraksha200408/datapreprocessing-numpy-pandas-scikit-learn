```python
import pandas as pd
import numpy as np

data = {
    "Product": ["Rice", "Wheat", "Maize", "Rice", "Ragi",
                "Wheat", "Maize", "Ragi"],
    "Category": ["Grain", "Grain", "Grain", "Grain",
                 "Millet", "Grain", "Grain", "Millet"],
    "Price": [2500, np.nan, 1800, 2500, 2200, 2100, np.nan, 2200],
    "Quantity": [10, 15, np.nan, 10, 8, 12, 20, 8],
    "Rating": [4.5, 4.0, np.nan, 4.5, 3.5, 4.2, 3.8, 3.5]
}

df = pd.DataFrame(data)


df=df.loc[(df["Price"]>2000) & (df["Rating"]>=4.0)]
df=df.drop_duplicates(subset=["Price" , "Rating"])
df=df.sort_values(by="Price", ascending=False)
print(df.loc[:,["Product", "Price", "Rating"]])
```

```python
import pandas as pd
import numpy as np

data = {
    "Product": ["Rice", "Wheat", "Maize", "Rice", "Ragi",
                "Wheat", "Maize", "Ragi"],
    "Category": ["Grain", "Grain", "Grain", "Grain",
                 "Millet", "Grain", "Grain", "Millet"],
    "Price": [2500, np.nan, 1800, 2500, 2200, 2100, np.nan, 2200],
    "Quantity": [10, 15, np.nan, 10, 8, 12, 20, 8],
    "Rating": [4.5, 4.0, np.nan, 4.5, 3.5, 4.2, 3.8, 3.5]
}

df = pd.DataFrame(data)

df["Stock_Value"] = df["Price"] * df["Quantity"]

result = df.sort_values(
    by="Price",
    ascending=False,
    na_position="last"
)

result = result.drop_duplicates(
    subset=["Product", "Category"],
    keep="first"
)

result = result.sort_values(
    by="Stock_Value",
    ascending=False,
    na_position="last"
)

print(result[[
    "Product", "Category", "Price",
    "Quantity", "Stock_Value"
]])
```
