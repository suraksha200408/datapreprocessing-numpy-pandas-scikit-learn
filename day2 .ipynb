```python
import numpy as np

sensor_data = np.array([
    [45.0, 120.0, 0.8, 210.0],
    [np.nan, 155.0, 2.4, 285.0],
    [38.0, np.inf, 0.5, 180.0],
    [81.0, 195.0, -np.inf, 365.0],
    [55.0, 140.0, 1.5, 250.0],
    [np.inf, 175.0, 3.1, np.nan],
    [42.0, 115.0, np.nan, 195.0],
    [76.0, 185.0, 3.7, 345.0]
])

#Create a Boolean array identifying all individual sensor values that are not finite. Print the Boolean array and the total number of invalid values.
con=~np.isfinite(sensor_data)
print(~np.isfinite(sensor_data),con.sum())
```

```python
#For each column, calculate how many NaN values it contains. Return one count per column.
print(np.sum(np.isnan(sensor_data),axis=0))
```

```python
contain = np.all(np.isfinite(sensor_data), axis=1)

clean_data = sensor_data[contain]

print(clean_data)
print(clean_data.shape)
```
