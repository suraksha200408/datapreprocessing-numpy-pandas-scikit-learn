```python
import pandas as pd
import numpy as np

data = {
    "Order_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "City": ["Bengaluru", "Mysuru", "Bengaluru", "Delhi", "Mysuru",
             "Delhi", "Bengaluru", "Delhi", "Mysuru", "Bengaluru"],
    "Restaurant_Type": ["Cafe", "FastFood", "Cafe", "FineDining", "FastFood",
                        "Cafe", "FineDining", "FastFood", "Cafe", "FastFood"],
    "Order_Value": [450, 300, 700, 1200, 250, 500, 1500, 350, 800, 400],
    "Delivery_Minutes": [25, 35, 30, 50, 28, 40, 55, 32, 38, 22],
    "Rating": [4.5, 3.8, 4.8, 4.2, 3.5, 4.0, 4.7, 3.9, 4.6, 4.1],
    "Discount": [50, 20, 100, 150, 0, 50, 200, 30, 80, 40]
}

df = pd.DataFrame(data)
df.groupby("City")["Order_Value"].agg(["sum", "mean","count"])
```

```python
result = df.groupby(
    ["City", "Restaurant_Type"],
    as_index=False
).agg(
    Total_Order_Value=("Order_Value", "sum"),
    Avg_Delivery_Minutes=("Delivery_Minutes", "mean")
)

print(result)
```
